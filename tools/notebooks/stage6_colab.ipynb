# Foehncraft – stage 6 on a Colab GPU

Runs the stage-6 parity check of the turbulence models (specification 6.6,
[docs/verification/stage6.md](https://github.com/chiefenne/Foehncraft/blob/main/docs/verification/stage6.md)):
every model (k-ω SST, standard and realizable k-ε, k-ε-ζ-f, k-ε-ζ-f₀, elliptic-blending k-ε)
